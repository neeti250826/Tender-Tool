```python
# @title Filters
DATE_FROM = '2026-03-01'  # @param {type:'string'}
DATE_TO = '2026-03-14'  # @param {type:'string'}
QUERY = ''  # @param {type:'string'}
BUYER_NAME = ''  # @param {type:'string'}
CPV_CODES = ''  # @param {type:'string'}
PAGE = 1  # @param {type:'integer'}
PAGE_SIZE = 50  # @param {type:'integer'}
DISCOVER_ONLY = False  # @param {type:'boolean'}
OUTPUT_TARGET = './_local_outputs'  # @param {type:'string'}
PROJECT_NAME = 'MDT_2026'  # @param {type:'string'}
WEBSITE_ID = 'RO_SEAP'  # @param {type:'string'}
SOURCE_LABEL = 'Romania SEAP'  # @param {type:'string'}
REGION = 'EMEA'  # @param ['EMEA','LATAM']
```

```python
# @title Advanced filters help
!python "ro_seap_scraper.py" --list-advanced-filters
```

```python
# Run (TED fallback by default). Set DISCOVER_ONLY=True to capture SEAP portal endpoints when reachable.
cpvs = [c.strip() for c in str(CPV_CODES or '').split(',') if c.strip()]
cpv_args = sum([['--cpv', c] for c in cpvs], [])
discover_args = ['--discover-only'] if DISCOVER_ONLY else []
cmd = [
  'python', 'ro_seap_scraper.py',
  '--date-from', DATE_FROM,
  '--date-to', DATE_TO,
  '--query', QUERY,
  '--buyer-name', BUYER_NAME,
  '--page', str(PAGE),
  '--page-size', str(PAGE_SIZE),
  '--output-target', OUTPUT_TARGET,
  '--project-name', PROJECT_NAME,
  '--website-id', WEBSITE_ID,
  '--source-label', SOURCE_LABEL,
  '--region', REGION,
] + cpv_args + discover_args
import subprocess
print(' '.join(cmd))
subprocess.run(cmd, check=False)
```
