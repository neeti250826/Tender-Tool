```python
# @title Config
DATE_FROM = '2026-03-01'  # @param {type:'string'}
DATE_TO = '2026-03-14'  # @param {type:'string'}
QUERY = ''  # @param {type:'string'}
OUTPUT_TARGET = './_local_outputs'  # @param {type:'string'}
PROJECT_NAME = 'MDT_2026'  # @param {type:'string'}
WEBSITE_ID = 'PL_EZAMOWIENIA'  # @param {type:'string'}
SOURCE_LABEL = 'Poland e-Zamowienia'  # @param {type:'string'}
REGION = 'EMEA'  # @param ['EMEA','LATAM']

NOTICE_TYPE = 'ContractNotice'  # @param {type:'string'}
BELOW_EU = True  # @param {type:'boolean'}
PAGE_NUMBER = 1  # @param {type:'integer'}
PAGE_SIZE = 0  # @param {type:'integer'}
SORTING_COLUMN = 'PublicationDate'  # @param {type:'string'}
SORTING_DIRECTION = 'DESC'  # @param ['ASC','DESC']
```

```python
# @title Advanced filters help
!python "ezamowienia_scraper.py" --list-advanced-filters
```

```python
import subprocess, sys
command = [sys.executable, 'ezamowienia_scraper.py',
    '--date-from', DATE_FROM, '--date-to', DATE_TO, '--query', QUERY,
    '--output-target', OUTPUT_TARGET, '--project-name', PROJECT_NAME,
    '--website-id', WEBSITE_ID, '--source-label', SOURCE_LABEL, '--region', REGION,
    '--notice-type', NOTICE_TYPE, '--page-number', str(PAGE_NUMBER),
    '--page-size', str(PAGE_SIZE), '--sorting-column', SORTING_COLUMN,
    '--sorting-direction', SORTING_DIRECTION]
if BELOW_EU:
    command.append('--below-eu')
subprocess.run(command, check=True)
```
